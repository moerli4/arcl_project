### Experiment 1

For the first experiment we consider three non conflicting cartesian task-space objectives.  
Task 1 is reaching to a desired X, task 2 is reaching to a desired Y and task 3 is to follow a sinusoidal trajectory in the Z coordinate.  
We do this with both the QP based controller and a traditional nullspace approach for comparison.  

This experiment shows that the implementation from the paper works well and is comparable to traditional approaches in tracking performance while achieving lower total command torques.  
Nevertheless it does require significantly more computational effort. It may be possible to improve this by using a different solver for the optimization.

Also note that epsilon provides a way to penalize applied command torque within the optimization. 

```python
# experiment parameters
show_viewer = False
T = 20
gravity = (0.0, 0.0, -9.81)

# define control objectives
tasks = [
    XPositionTask(x_des=0.4, K=20, D=10),
    YPositionTask(y_des=0.4, K=20, D=10),
    ZSinusoidalTask(z_center=0.4, amplitude=0.15, frequency=0.1, dt=dt, K=20, D=10),
]

# no artifical torque limits
torque_limits = None

# initialize results object
results = {}
```

```python
# QP based controller (as in the paper)
results["QP"] = run(
    tasks=tasks,
    controller=TorqueControllerQP(epsilon=1e-8),
    show_viewer=show_viewer,
    dt=dt,
    T=T,
    gravity=gravity,
    torque_limits=torque_limits,
)
```

```python
# traditional nullspace controller (as in the lecture)
results["Traditional"] = run(
    tasks=tasks,
    controller=TorqueControllerTraditional(),
    show_viewer=show_viewer,
    dt=dt,
    T=T,
    gravity=gravity,
    torque_limits=torque_limits,
)
```

```python
_ = plot(results=results, dt=dt, normalize_torque=False, sharey=True)
_ = show_metrics(results=results, dt=dt, transient_time=5)
```

### Experiment 2

In the second experiment we consider conflicting tasks to show that the qp does strictly hold the priority. 

For this we command the same task of reaching to an X position twice with a different value each.

It can be seen clearly from the residual task errors that task 1 reaches approximately zero error while task 2 does not. This is to be expected as task 2 directly conflicts with the higher priority task 1.

```python
# traditional nullspace controller (as in the lecture)
results["Traditional"] = run(
    tasks=tasks,
    controller=TorqueControllerTraditional(),
    show_viewer=show_viewer,
    dt=dt,
    T=T,
    gravity=gravity,
    torque_limits=torque_limits,
)
```

```python
_ = plot(results=results, dt=dt, normalize_torque=False, sharey=True)
_ = show_metrics(results=results, dt=dt, transient_time=5)
```

### Experiment 3

In the third experiment we consider the same tasks as in experiment 1. Additionally we introduce artificial torque limits in order to showcase the QPs capability of taking torque limit constraints into account directly.  
We also increase the gains of the tasks to ensure the torque limits do actually get reached.

For this experiment we do not simulate gravity as this makes it easier to see the direct effects of the torque limit on the tasks themselves.

It can be seen that the QP does achieve a lower maximum tracking error after transient time than the traditional controller. This is especially obvious at the times where the traditional approach reaches torque limits and saturates where the qp can directly consider these constraints (around 6s to 7.5s and 16s to 17s). Additionally the commanded torque RMS is lower in QP than traditional. This showcases the methods potential in improving tracking performance if torque limits constraints become active.

```python
# experiment parameters
show_viewer = False
T = 20
gravity = (0.0, 0.0, 0.0)

# define control objectives
tasks = [
    XPositionTask(x_des=0.4, K=100, D=50),
    YPositionTask(y_des=0.4, K=100, D=50),
    ZSinusoidalTask(z_center=0.4, amplitude=0.5, frequency=0.1, dt=dt, K=100, D=50),
]

# artificially limit torque
torque_limits = [8, 8, 8, 8, 8, 8, 8]

# initialize results object
results = {}
```

```python
# QP based controller (as in the paper)
results["QP"] = run(
    tasks=tasks,
    controller=TorqueControllerQP(epsilon=1e-8),
    show_viewer=show_viewer,
    dt=dt,
    T=T,
    gravity=gravity,
    torque_limits=torque_limits,
)
```

```python
# traditional nullspace controller (as in the lecture)
results["Traditional"] = run(
    tasks=tasks,
    controller=TorqueControllerTraditional(),
    show_viewer=show_viewer,
    dt=dt,
    T=T,
    gravity=gravity,
    torque_limits=torque_limits,
)
```

```python
_ = plot(results=results, dt=dt, normalize_torque=True, sharey=True)
_ = show_metrics(results=results, dt=dt, transient_time=5)
```
